```python
# INHERITANCE  

# single inheritance 
class Grandpa :      #    base class   / parrent class
    def sleep(self): 
        print("Grandpa is sleeping ")   


class Father(Grandpa): 
    def cook(self):  
        print("Father is cooking")       # child class      


f=Father()
f.cook()
f.sleep()
```

```python
g=Grandpa()
g.sleep()
```

```python
g.cook()
```

```python

# umltilevel inheritance  


class Herapheri:      # base class
    def laxmiprasad(self): 
        print(' wrong number yehi namah ')

class Phirherapheri(Herapheri):    # child class
    def  laxi(self): 
        print("21 din mai paisa double")   


class herapheri3(Phirherapheri): 
    def kachraseth(self): 
        print("150 rupaya dega ")  

h=herapheri3()
h.kachraseth()
h.laxi()
h.laxmiprasad()
```

```python
# Multiple inheritance   
class Horse: 
    def Power(self): 
        print("  HorsePower is Unit of Force ")   

class Donkey: 
    def Hardwork(self):   
        print(" Donkey is Considered as a Fool")

class Mule(Horse,Donkey): 
    def carrier(self):    
        print("Mule is use as a carrier at higher altitude")   


m=Mule()
m.carrier() 
m.Hardwork() 
m.Power()
```

```python
class addi: 
    def add(self,a,b): 
        return a+b   

class sub : 
    def s(self,a,b): 
        return a-b  

class multi : 
    def m(self,a,b): 
        return a*b  


class div(addi,sub,multi): 
    def d(self,a,b): 
        return a/b   


a=int(input("A="))
b=int(input("B="))  

di=div()
print(di.d(a,b) )
di.m(a,b)
```

```python
# Hiearchial inheritance    
class Grandfather: 
    def Money(self): 
        print("Paise hi Paise ")  

class Father(Grandfather): 
    def Gold(self):   
        print("Dad has Gold") 




class Uncle(Grandfather): 
    def Land(self):   
        print("Unncle has Land ")   

f=Father()
f.Gold()  

f.Money()
```

```python
u=Uncle()
u.Land()
u.Money()
```
